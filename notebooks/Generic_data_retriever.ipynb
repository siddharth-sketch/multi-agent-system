```python
import random
import uuid
import pandas as pd
from faker import Faker
from datetime import datetime, timedelta

fake = Faker()

# Sample values
product_types = ["Mobile", "Internet", "WiFi", "DTH"]
product_names = {
    "Mobile": ["SuperNet 299", "Etisalat Flexi", "TalkMax", "Roaming Saver"],
    "Internet": ["Fiber Plus 100", "HomeNet", "SpeedBoost", "Unlimited Surf"],
    "WiFi": ["WiFi Pro", "SmartHub", "Zone Booster", "Mesh Ultra"],
    "DTH": ["TV Max", "Entertainment Combo", "Sports Pack", "Kids World"]
}
payment_methods = ["Card", "UPI", "Wallet", "Netbanking"]
regions = ["Dubai", "Abu Dhabi", "Sharjah", "Ajman", "Fujairah"]
channels = ["App", "Website", "Store"]

# Generate 100 rows
sales_data = []
for _ in range(100):
    product_type = random.choice(product_types)
    product_name = random.choice(product_names[product_type])
    price = round(random.uniform(29.0, 499.0), 2)
    discount = round(random.uniform(0.0, 50.0), 2) if random.random() < 0.3 else 0.0
    
    sales_data.append({
        "sale_id": str(uuid.uuid4()),
        "customer_id": str(uuid.uuid4()),
        "product_type": product_type,
        "product_name": product_name,
        "sale_date": fake.date_between(start_date='-6M', end_date='today'),
        "price": price,
        "payment_method": random.choice(payment_methods),
        "region": random.choice(regions),
        "channel": random.choice(channels),
        "is_recurring": random.choice([True, False]),
        "discount_applied": discount,
        "delivery_status": random.choice(["Pending", "Delivered", "Failed"])
    })

sales_df = pd.DataFrame(sales_data)
sales_df.head()
```

```python
# Use the existing sales_df to link delivery.sale_id to sales.sale_id
delivery_data = []

for i in range(100):
    sale_id = sales_df.iloc[i]["sale_id"]
    
    delivery_status = random.choice(["Pending", "Delivered", "Failed"])
    delivery_type = random.choice(["Digital", "Physical", "Remote"])
    failure_reason = None if delivery_status != "Failed" else random.choice([
        "Customer unavailable", "Address not found", "Technical issue", "Payment pending"
    ])
    delivery_date = (
        sales_df.iloc[i]["sale_date"] + timedelta(days=random.randint(0, 5))
        if delivery_status in ["Delivered", "Failed"] else None
    )
    delivery_time = f"{random.randint(9, 11)}AM - {random.randint(1, 3)}PM" if delivery_status != "Pending" else None
    rating = random.randint(1, 5) if delivery_status == "Delivered" and random.random() < 0.5 else None
    
    delivery_data.append({
        "delivery_id": str(uuid.uuid4()),
        "sale_id": sale_id,
        "delivery_type": delivery_type,
        "assigned_agent": fake.name() if delivery_type == "Physical" else "System Bot",
        "delivery_date": delivery_date,
        "delivery_time": delivery_time,
        "delivery_status": delivery_status,
        "failure_reason": failure_reason,
        "delivery_cost": round(random.uniform(0.0, 20.0), 2),
        "retry_attempts": random.randint(0, 3) if delivery_status == "Failed" else 0,
        "rating": rating,
        "delivery_notes": fake.sentence(nb_words=6) if random.random() < 0.3 else None
    })

delivery_df = pd.DataFrame(delivery_data)
delivery_df.head()
```

```python
# Generate 100 unique customer IDs (to match with sales data)
unique_customer_ids = sales_df["customer_id"].unique().tolist()

# Gender and language options
genders = ["M", "F", "Other"]
languages = ["English", "Arabic", "Hindi", "Urdu", "Tagalog"]

customer_data = []
for cust_id in unique_customer_ids:
    dob = fake.date_of_birth(minimum_age=18, maximum_age=65)
    customer_data.append({
        "customer_id": cust_id,
        "full_name": fake.name(),
        "email": fake.email(),
        "phone_number": fake.phone_number(),
        "registration_date": fake.date_between(start_date='-2y', end_date='today'),
        "dob": dob,
        "gender": random.choice(genders),
        "region": random.choice(regions),
        "preferred_language": random.choice(languages),
        "is_active": random.choice([True, False]),
        "account_type": random.choice(["Prepaid", "Postpaid"])
    })

customer_df = pd.DataFrame(customer_data)
customer_df.head()
```

```python
import sqlite3

# Connect to an in-memory SQLite database
conn = sqlite3.connect("etisalat_sales.db")

# # Write the DataFrames to SQLite tables
sales_df.to_sql("sales", conn, if_exists="replace", index=False)
delivery_df.to_sql("delivery", conn, if_exists="replace", index=False)
customer_df.to_sql("customer", conn, if_exists="replace", index=False)

# List all tables to confirm
cursor = conn.cursor()
cursor.execute("SELECT * FROM delivery WHERE sale_id IN (SELECT sale_id FROM sales WHERE payment_method = 'UPI');")
tables = cursor.fetchall()
conn.commit()
conn.close()

print(tables)
```
